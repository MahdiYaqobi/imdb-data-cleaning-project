# IMDb Movies EDA and Visualization

This notebook performs basic analysis, exploratory data analysis (EDA), and visualizations on the cleaned IMDb dataset.

```python
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
pd.set_option('display.max_columns', 200)
```

## Load the Cleaned Dataset

Load the cleaned IMDb dataset from the processed CSV file.

```python
# Load the cleaned IMDb dataset
imdb_df = pd.read_csv('../data/processed/imdb_clean.csv')
```

## Display Dataset Shape and Columns

Show the shape and column names of the dataset.

```python
# Show shape and columns
print('Shape:', imdb_df.shape)
print('Columns:', imdb_df.columns.tolist())
```

## Summary Statistics for Numerical Columns

Display summary statistics for all numerical columns.

```python
# Summary statistics for numerical columns
imdb_df.describe()
```

## Distribution of IMDb Ratings

Plot a histogram and KDE of the 'imdb_rating' column.

```python
# Histogram and KDE of IMDb Ratings
plt.figure(figsize=(8,5))
sns.histplot(imdb_df['imdb_rating'], kde=True, bins=20, color='skyblue')
plt.title('Distribution of IMDb Ratings')
plt.xlabel('IMDb Rating')
plt.ylabel('Count')
plt.show()
```

## Distribution of Gross Revenue

Plot a histogram of the 'gross' column.

```python
# Histogram of Gross Revenue
plt.figure(figsize=(8,5))
sns.histplot(imdb_df['gross'], bins=30, color='orange')
plt.title('Distribution of Gross Revenue')
plt.xlabel('Gross Revenue')
plt.ylabel('Count')
plt.show()
```

## Movies Count by Decade

Plot a bar chart showing the number of movies released in each decade.

```python
# Bar chart: Movies count by decade
plt.figure(figsize=(8,5))
sns.countplot(x='decade', data=imdb_df, color='teal')
plt.title('Number of Movies by Decade')
plt.xlabel('Decade')
plt.ylabel('Number of Movies')
plt.show()
```

## Movies Count by Released Period

Plot a countplot for the 'released_period' column.

```python
# Countplot for released_period
plt.figure(figsize=(6,4))
sns.countplot(x='released_period', data=imdb_df, palette='Set2')
plt.title('Movies Count by Released Period')
plt.xlabel('Released Period')
plt.ylabel('Number of Movies')
plt.show()
```

## Movies Count by Main Genre

Plot the number of movies for each main genre.

```python
# Bar chart: Movies count by main genre
plt.figure(figsize=(10,6))
sns.countplot(y='main_genre', data=imdb_df, order=imdb_df['main_genre'].value_counts().index, palette='viridis')
plt.title('Number of Movies by Main Genre')
plt.xlabel('Number of Movies')
plt.ylabel('Main Genre')
plt.show()
```

## Boxplot of IMDb Ratings by Genre

Create a boxplot to compare IMDb ratings across main genres.

```python
# Boxplot: IMDb ratings by main genre
plt.figure(figsize=(12,6))
sns.boxplot(x='main_genre', y='imdb_rating', data=imdb_df, palette='Set3')
plt.title('IMDb Ratings by Main Genre')
plt.xlabel('Main Genre')
plt.ylabel('IMDb Rating')
plt.xticks(rotation=45)
plt.show()
```

## Correlation Heatmap

Plot a heatmap of the correlation matrix for numerical features.

```python
# Correlation heatmap
plt.figure(figsize=(10,8))
corr = imdb_df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

## Scatter Plot: Gross vs IMDb Rating

Create a scatter plot to visualize the relationship between gross revenue and IMDb rating.

```python
# Scatter plot: Gross vs IMDb Rating
plt.figure(figsize=(8,6))
sns.scatterplot(x='imdb_rating', y='gross', data=imdb_df, alpha=0.7)
plt.title('Gross Revenue vs IMDb Rating')
plt.xlabel('IMDb Rating')
plt.ylabel('Gross Revenue')
plt.show()
```

## Top 10 Movies by Gross Revenue

Display and plot the top 10 movies with the highest gross revenue.

```python
# Top 10 movies by gross revenue
top_gross = imdb_df.nlargest(10, 'gross')
display(top_gross[['series_title', 'gross', 'imdb_rating']])
plt.figure(figsize=(10,6))
sns.barplot(y='series_title', x='gross', data=top_gross, palette='Blues_r')
plt.title('Top 10 Movies by Gross Revenue')
plt.xlabel('Gross Revenue')
plt.ylabel('Movie Title')
plt.show()
```

## Top 10 Movies by IMDb Rating

Display and plot the top 10 movies with the highest IMDb ratings.

```python
# Top 10 movies by IMDb rating
top_rating = imdb_df.nlargest(10, 'imdb_rating')
display(top_rating[['series_title', 'imdb_rating', 'gross']])
plt.figure(figsize=(10,6))
sns.barplot(y='series_title', x='imdb_rating', data=top_rating, palette='Greens_r')
plt.title('Top 10 Movies by IMDb Rating')
plt.xlabel('IMDb Rating')
plt.ylabel('Movie Title')
plt.show()
```

## Pairplot of Key Numerical Features

Create a pairplot of selected numerical features such as imdb_rating, gross, runtime, and no_of_votes.

```python
# Pairplot of key numerical features
sns.pairplot(imdb_df[['imdb_rating', 'gross', 'runtime', 'no_of_votes']].dropna())
plt.suptitle('Pairplot of Key Numerical Features', y=1.02)
plt.show()
```
